# AI-Based Network Intrusion Detection System
## Notebook 02: Model Training, Benchmarking & Evaluation
**Objective**: Evaluate Logistic Regression (Baseline), Random Forest, and XGBoost on official UNSW-NB15 test partition.
- Primary Task: Binary Classification (Normal vs Attack)
- Secondary Task: Multiclass Classification (10 Attack Categories)

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

with open("../models/training_results.json", "r") as f:
    results = json.load(f)

print("Benchmark summary loaded successfully.")
```

### 1. Binary Classification Model Comparison (Primary Task)

```python
binary_data = []
for name, exp in results["binary_experiments"].items():
    m = exp["metrics"]
    binary_data.append({
        "Model": name,
        "Accuracy (%)": round(m["accuracy"] * 100, 2),
        "Precision (%)": round(m["precision"] * 100, 2),
        "Recall (%)": round(m["recall"] * 100, 2),
        "F1-Score": m["f1"],
        "ROC-AUC": m["roc_auc"],
        "Train Time (s)": exp["training_time_seconds"],
        "Peak RAM (MB)": exp["peak_memory_mb"],
        "Latency (ms/sample)": exp["inference_latency"]["per_sample_latency_ms"]
    })
df_binary = pd.DataFrame(binary_data)
display(df_binary)
```

### 2. Multiclass Classification Model Comparison (10 Categories)

```python
multi_data = []
for name, exp in results["multiclass_experiments"].items():
    m = exp["metrics"]
    multi_data.append({
        "Model": name,
        "Accuracy (%)": round(m["accuracy"] * 100, 2),
        "Macro F1": m["f1_macro"],
        "Weighted F1": m["f1_weighted"],
        "Macro ROC-AUC": m["roc_auc_macro"],
        "Train Time (s)": exp["training_time_seconds"],
        "Peak RAM (MB)": exp["peak_memory_mb"]
    })
df_multi = pd.DataFrame(multi_data)
display(df_multi)
```

### 3. Per-Class Analysis for Rare Attacks (Worms & Shellcode)
Analysis of XGBoost (Multiclass) performance on the long-tailed classes:

```python
xgb_report = results["multiclass_experiments"]["XGBoost (Multiclass)"]["metrics"]["classification_report"]
df_report = pd.DataFrame(xgb_report).transpose()
display(df_report.loc[["Generic", "Normal", "Reconnaissance", "Exploits", "Shellcode", "Worms", "Analysis", "Backdoor"]])
```
